Greate a dict

```python
weeks = {
    "1" : "Monday",
    "2" : "Tuesday",
    "3" : "Wednesday",
    "4" : "Thursday",
    "5" : "Friday",
    "6" : "Saturday",
    "7" : "Sunday",
}
print(weeks)
```

Accessing Items

Key

```python
print(weeks['1'])
```

```python
print(weeks['3'])
```

get()

```python
print(weeks.get("9","Гараг нь зөвхөн 7 өдрөөс бүрдэнэ!"))
```

Change value

```python
weeks['1'] = "Mon"
```

```python
print(weeks)
```

Loop through a dict

key

```python
for x in weeks:
  print(x)
```

```python
for x in weeks.keys():
  print(x)
```

value

```python
for x in weeks.values():
  print(x)
```

items

```python
for x, y in weeks.items():
  print(x,y)
```

```python
for d in weeks.items():
  print(d)
```

Membership

```python
if '1' in weeks:
  print("YES")
```

```python
if 1 in weeks:
  print("YES")
else:
  print("NO")
```
