```python
a = 10
b = 30
if a < b:
  print("a baga")
else:
  print("b baga")
```

```python
a, b = 10, 5

print("a baga" if a < b else "b baga")
```

List Comprehension

```python
[i ** 2 for i in range(1,11)]
```

```python
[i for i in range(1,100) if i % 11 == 0 or i % 23 == 0]
```

```python
[(i,"even") if i % 2 == 0 else (i, "odd") for i in range(1,11)]
```

```python
[x * y for x in range(1,4) for y in range(3,6)]
```

```python
x = int(input())
y = int(input())
n = int(input())
print([[i,j] for i in range(0, x+1) for j in range(0, y+1) if (i + j) != n])
```
