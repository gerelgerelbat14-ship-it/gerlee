```python
from string import ascii_lowercase, ascii_uppercase


def solve(s):
    for c in s:
        if c in ascii_lowercase or c in ascii_uppercase:
            return True
        else:
            return False
```

```python
"".join(list(filter(solve, "The i notebook wep #41 sok.")))
```

```python
list(filter(lambda x : x % 2 == 0, [1, 11, 20, 30, 49, 99, 100]))
```

```python
"".join(
    list(
        filter(
            lambda s: s.lower() == "!",
            "The ! notebtheook dafok peoprothe[elf[hagiugjikljdahashdljhaldkjhfhqfh;dioufgwep #41 sok.",
        )
    )
)
```
