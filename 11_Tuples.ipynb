Tuples

Нэг үүссэн бол доторх элементийг өөрчилж болохгүй, дараалласан бүтэцтэй.

Greate a Tuple

```python
my_tuples = ("iphone", "samsung", "huawei", "nokia")
print(my_tuples)
```

Access Tuple Items

```python
print(my_tuples[1])
```

Change Tuples Values

```python
my_tuples[1] = 'smsng'
```

Loop through a tuple

```python
for x in my_tuples:
  print(x)
```

Membership oprator

```python
if 'huawei' in my_tuples:
  print("Yes, huawei in your tuples")
else:
    print("Nothing")
```

Length

```python
len(my_tuples)
```

Add Items

```python
my_tuples[4] = 'ip'
```

```python
my_tuples.insert(0,"ip")
```

Remove Items

```python
print(my_tuples)
del my_tuples
print(my_tuples)
```
