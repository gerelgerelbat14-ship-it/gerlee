#### Бодлого

Гараас өгсөн list доторх элементийг тоолдог код бич

```python
arr = list(map(int, input().split()))

count_dict = {}

for el in arr:
  if el in count_dict:
    count_dict[el] += 1
else:
      count_dict[el] = 1

print(count_dict)
```

```python
for k in count_dict:
  print('{} элемент {} удаа орсон байна.'.format(k, count_dict[k]))
```

#### While

!! Заавал код зогсох нөхцөлтэй байх ёстой

```python
i = 1

while i < 11:
  print(i)
  i += 1
```

```python
lst = [1, 1]

i = 2

while i < 10:
  lst.append(lst[-1] + lst[-2])
  i += 1

print(lst)
```
